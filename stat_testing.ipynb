```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy import stats
from scipy.stats import ttest_ind, chi2_contingency
from statsmodels.stats.proportion import proportions_ztest
```

```python
df= pd.read_csv('marketing_AB.csv')
df.head()
```

```python
df['test group'].value_counts()
df.groupby('test group')['converted'].mean()
```

```python
ad_group = df[df['test group']== 'ad']
psa_group = df[df['test group']== 'psa']

converted_ad = ad_group['converted'].sum()
converted_psa = psa_group['converted'].sum()
n_ad = ad_group['converted'].count()
n_psa = psa_group['converted'].count()
count = [converted_ad, converted_psa]
nop = [n_ad,n_psa]
z_stat,p_value = proportions_ztest(count,nop,alternative='larger')
z_stat,p_value
```
