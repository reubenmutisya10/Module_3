# Cloud Computing
Cloud Computing - Refers to the use of remote servers, storage, and computing power accessible via the internet, rather than relying on local machines or on-premises infrastructure. 

Cloud computing enables data scientists to access scalable remote storage and computing resources over the internet, making it easier to process large datasets, build complex machine learning models, collaborate, and scale projects. Major platforms like AWS, Microsoft Azure, and Google Cloud Platform (GCP) provide specialized tools that support efficient data science workflows.

## Example: Healthcare - Predicting Patient Readmissions
A data scientist at a large hospital network wants to predict which patients are likely to be readmitted after discharge.

### Data Storage
Patient records, lab results, medical histories, and hospital visits are securely stored in cloud storage instead of local computers.

### Data Processing
Large volumes of healthcare data are cleaned, merged, and transformed using cloud-based processing tools.

### Model Training
A machine learning model is trained in the cloud to identify patients at high risk of readmission without requiring expensive local hardware.

### Model Deployment
The trained model is deployed as an online service that provides real-time risk predictions whenever a patient is discharged.

### Collaboration and Scaling
Doctors, analysts, and data scientists across different hospitals can access shared resources, while the cloud automatically scales as more hospitals and patient data are added.

- Cloud computing reduces infrastructure management, allowing data scientists to focus on solving real-world problems and building better machine learning solutions.

## Key Cloud Computing Terms in Data Science
- **On-Premises Computing:** A traditional setup where an organization owns and manages its own servers and infrastructure.
- **Cloud Platform:** A service that provides computing resources over the internet (e.g., AWS, Microsoft Azure, Google Cloud).
- **Scalability:** The ability to increase or decrease computing resources based on demand.
- **On-Demand Resources:** Computing resources that can be provisioned and released whenever they are needed.
- **Agility:** The ability to quickly adapt to changing workloads and business needs.
- **Performance:** The efficient processing of data and execution of applications using cloud resources.

## How Does the Cloud Work for Data Scientists?
- **Infrastructure as a Service (IaaS)**
Infrastructure as a Service (IaaS) provides virtualized computing resources such as servers, storage, and networking over the internet. Instead of purchasing and maintaining physical hardware, developers can create, test, and deploy applications using cloud-based infrastructure. Resources can be increased or reduced as project requirements change, making development more flexible and cost-effective.

- **Scalable Storage and Processing**
Cloud platforms provide storage and computing resources that automatically adapt to the workload. When processing large datasets or training machine learning models, additional computing power can be allocated as needed. Once the workload decreases, the extra resources are released, helping organizations optimize both performance and cost.

- **APIs and Cloud Services Integration**
Cloud providers offer APIs and managed services that developers can integrate directly into their applications. Using programming languages such as Python, developers can connect to cloud storage, databases, machine learning services, and other cloud resources without having to build these systems from scratch. This accelerates application development and simplifies deployment.

- **Software as a Service (SaaS)**
Software as a Service (SaaS) delivers software applications through the internet, allowing users to access them using a web browser. Developers and data scientists commonly use cloud-hosted tools such as Jupyter Notebooks, GitHub, and collaborative development platforms to write code, manage projects, and work with team members without installing or maintaining software on their local machines.

- **Automated Scaling and Performance Management**
Cloud platforms continuously monitor application performance and automatically adjust computing resources based on current demand. During periods of high traffic or heavy processing, additional resources are allocated to maintain performance. When demand decreases, resources are scaled down automatically, ensuring applications remain reliable while minimizing operational costs.

## 1. **Deploying a Model as an API**
After training a machine learning model, you can save it as a **pickle file** (`.pkl`) using Python.

A pickled model stores the trained model so that it can be loaded later without training it again.

##### The Limitation of Pickled Models
A pickled model is a Python file.
This means:
- It can only be loaded using Python.
- The user needs Python installed.
- The user needs the required libraries (such as scikit-learn).
- Other applications like websites or mobile apps cannot use the `.pkl` file directly.

##### The Solution: APIs
Instead of giving people the pickle file, you can build an **API (Application Programming Interface).**

An API acts as a **bridge** between an application and your machine learning model.

Instead of accessing the model directly, applications send data to the API, and the API returns the prediction.

The application does **not** need to know how the model works or even that it was built in Python.

##### How an API Works
The process is simple:

1. A client sends data to the API.
2. The API receives the data.
3. The API loads the trained model.
4. The model makes a prediction.
5. The API sends the prediction back to the client.

```
Client
   │  Send data
   ▼
API Server
   │  Uses trained model
   ▼
Machine Learning Model
   │  Prediction
   ▼
API Server
   │  Sends response
   ▼
Client
```

##### Example
A real estate company has built a machine learning model that predicts the selling price of a house.

The model has already been trained and is running on a server.

When someone visits the company's website and enters the details of a house, the website does **not** calculate the price itself. Instead, it sends the information to the API.

##### Step 1: The User Enters House Details
A user fills out a form on the website:

- Bedrooms: 3
- Bathrooms: 2
- House Size: 1,800 sq ft
- Location: Nairobi
- Age of House: 5 years

The user clicks **"Predict Price."**

##### Step 2: The Website Sends a Request
The website sends the house details to the API.

##### Step 3: The API Uses the Machine Learning Model
The API receives the house details.

Behind the scenes, it:

1. Receives the request.
2. Loads the trained machine learning model.
3. Passes the house details to the model.
4. The model predicts the selling price.

##### Step 4: The API Sends Back the Prediction
The server responds with the predicted price.

The model may predict that the house is worth **KES 12,500,000**.

##### Step 5: The Website Displays the Result
The website receives the response and displays:

> **Estimated House Price: KES 12,500,000**

##### Understanding the Request
A **request** is the information sent **to** the API.

In this example, the request contains the house details.

```http
POST /predict-price
```
```json
{
    "bedrooms": 3,
    "bathrooms": 2,
    "size_sqft": 1800,
    "location": "Nairobi",
    "house_age": 5
}
```
The request is asking:

> "Based on these house details, what is the estimated selling price?"
##### Understanding the Response
A **response** is the information sent **back** by the API after the prediction is made.

```json
{
    "predicted_price": 12500000
}
```

The website reads this response and shows the predicted price to the user.
##### Who Does What?

| Component | Responsibility |
|-----------|----------------|
| **User** | Enters the house details. |
| **Website (Client)** | Sends the house details to the API. |
| **API (Server)** | Receives the request and passes the data to the machine learning model. |
| **Machine Learning Model** | Predicts the house price. |
| **API** | Returns the prediction to the website. |
| **Website** | Displays the predicted price to the user. |

- The **website** sends data to the API.
- The **API** sends the data to the machine learning model.
- The **model** makes a prediction.
- The **API** returns the prediction.
- The **website** displays the result to the user.

##### Who is the Client?
A **client** is anything that requests predictions from the API.

Examples include:

- A website
- A mobile application
- Another Python program
- A desktop application
- An IoT device
##### Who is the Server?
The **server** is the computer running your machine learning model.

Its job is to:

- Receive requests
- Load the trained model
- Make predictions
- Send responses
##### Ways to Send Requests to an API
Many tools can communicate with an API.

Examples include:

- Python (`requests` library)
- JavaScript (`fetch()`)
- cURL (Terminal)
- Postman

All these tools can send requests to the same API.
##### Why Use an API?
Using an API offers several advantages:

- Makes the model available to applications written in different programming languages.
- Allows websites and mobile apps to use the model.
- Keeps the model on a central server instead of distributing it to every user.
- Makes updating the model easier because only the server needs to be updated.
- Enables multiple users to access the model at the same time.

## 2. **Deploying a Model as an API with Cloud Functions**
##### What is a Cloud Function?
A **cloud function** is a small piece of code that runs in the cloud whenever it receives a request.

Cloud functions are commonly used to deploy simple machine learning models because they require very little code.
##### Why Use a Cloud Function?
Suppose you have already trained a machine learning model and saved it as a file (`model.pkl`).

Instead of sending the model file to every user, you can upload it to the cloud along with a cloud function.

Whenever someone sends data to the cloud function:

1. The cloud function loads the model.
2. The model makes a prediction.
3. The cloud function returns the prediction.

Everything happens automatically.
##### How a Cloud Function Works
```
User/Application
       │ Sends request
       ▼
Cloud Function
       │ Loads model.pkl
       ▼
Machine Learning Model
       │ Makes prediction
       ▼
Cloud Function
       │ Sends prediction
       ▼
User/Application
```
##### Advantages of Cloud Functions
- Very little code is required.
- No need to manage servers.
- Automatically scales to handle many requests.
- Cost-effective because you only pay when the function runs.
- Easy to integrate with websites and mobile apps.
##### Limitations of Cloud Functions
- Initial configuration can be challenging.
- Debugging errors may take time since the code runs in the cloud.
- Functions have limits on execution time and memory.
- Large machine learning models may not fit within cloud function limits.
##### Cloud Function Providers
Several cloud providers offer cloud functions:
- **Google Cloud Functions**
- **AWS Lambda**
- **Azure Functions**
##### Real-World Example
A bank has built a model to detect fraudulent credit card transactions.

Instead of installing the model on every ATM or banking app:

1. The model is uploaded to the cloud.
2. Every time a transaction occurs, the banking app sends the transaction details to a cloud function.

## 3. **Deploying a Model as an API with Flask**
#### What is Flask?
**Flask** is a lightweight Python web framework used to build web applications and APIs.

It is called a **microframework** because it provides only the essential tools needed to create web applications, making it simple and easy to learn.

In machine learning, Flask is commonly used to deploy trained models as APIs so that websites, mobile applications, and other software can request predictions.

##### Why Use Flask?
After training a machine learning model, you want other applications to use it.

Instead of sharing the model file (`student_model.pkl`), you can create a Flask application that:

- Loads the trained model.
- Receives data from users.
- Uses the model to make predictions.
- Returns the prediction as a response.

This makes the model accessible over the internet.
##### How Flask Works
```
Website / Mobile App
         │ Sends Request
         ▼
     Flask API
         │ Loads student_model.pkl
         ▼
Machine Learning Model
         │ Makes Prediction
         ▼
     Flask API
         │ Sends Response
         ▼
Website / Mobile App
```
##### Example: Student Exam Pass Prediction API
Suppose a school has trained a machine learning model that predicts whether a student is likely to **Pass** or **Fail** an exam based on:

- Hours studied
- Attendance percentage

Instead of giving every teacher the machine learning model, the school deploys it using Flask.

Teachers simply send student information to the API and receive a prediction.
##### Flask Code
```python
from flask import Flask, request
import joblib
import json

app = Flask(__name__)

# Function that loads the model and predicts
def predict_pass(hours_studied, attendance):

    with open("student_model.pkl", "rb") as file:
        model = joblib.load(file)

    X = [[hours_studied, attendance]]

    prediction = model.predict(X)

    return {"prediction": prediction[0]}


# Home page
@app.route("/", methods=["GET"])
def home():
    return "Student Performance Prediction API"


# Prediction endpoint
@app.route("/predict", methods=["POST"])
def predict():

    data = request.get_json()

    result = predict_pass(
        data["hours_studied"],
        data["attendance"]
    )

    return json.dumps(result)
```

The machine learning model analyzes the student's study hours and attendance and predicts whether the student is likely to pass.
##### Understanding Flask Routes
A **route** is a URL that tells Flask what to do when someone visits it.
Example:

```python
@app.route("/")
```

This creates the home page.

Example:

```python
@app.route("/predict")
```

This creates the prediction endpoint.
#### GET vs POST Requests
##### GET Request
Used to retrieve information.
Example:

```
GET /
```
Response:
```
Student Performance Prediction API
```
##### POST Request
Used to send information to the server.
Example:
```
POST /predict
```
Request:

```json
{
    "hours_studied": 8,
    "attendance": 95
}
```

Response:

```json
{
    "prediction": "Pass"
}
```

- Lightweight and easy to learn.

Big Data refers refers to massive, complex, and rapidly growing datasets characterized by the three Vs:

- **Volume** The massive amount of data generated from various sources (e.g., social media, sensors, transactions).
- **Velocity** The speed at which data is generated, collected, and processed (often in real-time).

## Big Data: Key Concepts
### 1. MapReduce
A programming model used to process large datasets by dividing the work across many computers.

It is a core component of the Apache Hadoop framework, which enables the parallel processing of big data across clusters of computers.

How it works:
**Map Phase**: Splits a large dataset into smaller pieces and processes them in parallel.

**Reduce Phase**: Combines the processed results to produce the final output.

### 2. Resilient Distributed Dataset (RDD)
The basic data structure in Apache Spark used to store and process data across multiple computers.

Key Features:

**Resilient**: Can recover lost data if a computer fails.
**Distributed**: Data is split into partitions and stored across different machines.
**Dataset**: Can store both structured and unstructured data.

Key Benefit:

Enables very fast data processing because Spark keeps data in memory instead of repeatedly reading from disk.

### 3. Hadoop Distributed File System (HDFS)
A distributed storage system used by Hadoop to store very large datasets.

Key Features:

Stores data across many computers.
Creates multiple copies (replicas) of data for backup.
Continues working even if some machines fail.

Key Benefit:

Provides reliable, scalable, and fault-tolerant storage for big data.

### 4. Apache Spark
A powerful framework for processing big data quickly using in-memory computing.

What Spark can do:

Batch processing
Real-time data streaming
Machine learning
Graph processing

Why Spark is faster than Hadoop MapReduce:

Processes data in memory rather than reading and writing to disk repeatedly.

Spark is an evolution of Hadoop, offering higher speeds through in-memory processing and a more user-friendly API.

## How Does Big Data Work?
Big Data works by using distributed computing frameworks that process and analyze massive datasets across multiple computers (called nodes) at the same time.

Instead of relying on a single computer, Big Data systems divide the work among many machines, making data processing much faster and more efficient.

Why is this important?

Traditional computers struggle with:

- Extremely large datasets (terabytes or petabytes)
- Data generated very quickly (real-time data)
- Complex data from different sources

Big Data frameworks solve these problems through parallel processing and distributed computing.

### 1. Distributed Computing
Distributed computing is a method of processing data where many computers work together as one system.

Each computer (called a **node**) performs part of the work, and together they complete the task much faster than a single machine.

#### How it Works
Instead of:

```text
One Computer
      │
      ▼
Processes All Data
      │
      ▼
Very Slow
```

Big Data uses:

```text
Large Dataset
      │
      ▼
Split into Smaller Chunks
      │
      ▼
 ┌────────┬────────┬────────┬────────┐
 │ Node 1 │ Node 2 │ Node 3 │ Node 4 │
 └────────┴────────┴────────┴────────┘
      │       │        │       │
      ▼       ▼        ▼       ▼
 Process Process  Process Process
      │       │        │       │
      └───────┴────────┴───────┘
              │
              ▼
        Final Output
```
#### Benefits
- Faster processing
- Handles larger datasets
- Easily scalable
- Reliable even if one computer fails

### 2. Parallel Processing
Parallel processing is the technique of dividing a large task into multiple smaller tasks that run simultaneously on different computers.

Instead of processing data one record at a time:

```text
Record 1
Record 2
Record 3
Record 4
```

The system processes:

```text
Node 1 → Record 1
Node 2 → Record 2
Node 3 → Record 3
Node 4 → Record 4

(All processed simultaneously)
```

This significantly reduces processing time.

### 3. Parallel Processing in Hadoop (MapReduce)
In Hadoop, developers write two functions:

#### Map Function
The **Map** function:

- Reads the data
- Splits it into key-value pairs
- Processes each piece independently

#### Example
**Input**

```text
Apple Banana Apple Orange Banana Apple
```
**Map Output**

```text
(Apple,1)
(Banana,1)
(Apple,1)
(Orange,1)
(Banana,1)
(Apple,1)
```
#### Reduce Function
The Reduce function groups identical keys together and performs calculations.

#### Output

```text
Apple  → 3
Banana → 2
Orange → 1
```

#### MapReduce Workflow

```text
Input Data
     │
     ▼
 Map Function
     │
     ▼
Intermediate Key-Value Pairs
     │
     ▼
 Shuffle & Sort
     │
     ▼
 Reduce Function
     │
     ▼
 Final Result
```
### 4. Parallel Processing in Apache Spark

Apache Spark simplifies parallel processing by providing built-in operations called **transformations**.

#### Common Spark Transformations

| Transformation | Purpose |
|---------------|---------|
| `map()` | Modify every record |
| `filter()` | Keep records matching a condition |
| `flatMap()` | Split records into multiple outputs |
| `reduceByKey()` | Aggregate values with the same key |
| `groupByKey()` | Group records with the same key |

### 5. Fault Tolerance
Fault tolerance is the ability of a Big Data system to continue working even when one or more computers fail.

Since clusters contain hundreds or thousands of computers, failures are expected. Big Data frameworks automatically recover without stopping the entire system.
#### Fault Tolerance in Hadoop

Hadoop uses the **Hadoop Distributed File System (HDFS)**.

HDFS stores multiple copies (replicas) of every data block.
#### Example
```text
Block A

Stored on:
- Node 1
- Node 4
- Node 8
```

If **Node 1** fails:

```text
System automatically reads Block A from Node 4.
```

No data is lost.
#### Fault Tolerance in Spark

Spark uses **RDD Lineage**.

Instead of storing multiple copies of processed data, Spark remembers how every dataset was created.

If one partition is lost, Spark recomputes only the missing partition rather than rerunning the entire job.

## Big Data Process Explained with MapReduce

### 1. What is MapReduce?
MapReduce is a framework that splits a large problem into smaller tasks, processes them in parallel, and combines the results into one final output.

### Why is MapReduce Needed?
Imagine you have:

- 5 billion Google search queries
- 2 billion Facebook posts
- 500 million online shopping transactions

Processing all this data on one computer would:

- Take days or weeks.
- Consume huge amounts of memory.
- Risk failure if the computer crashes.

Instead, MapReduce distributes the work across hundreds or thousands of machines.

Example:

Instead of:
```text
1 Computer
Processes 1 Billion Records
(Time = 24 Hours)
```
MapReduce does:

```text
100 Computers

Each processes
10 Million Records

(Time = Minutes)
```
### How MapReduce Works
MapReduce consists of **three main stages**:

```text
Large Dataset
      │
      ▼
 1. MAP
      │
      ▼
Intermediate Key-Value Pairs
      │
      ▼
2. SHUFFLE
      │
      ▼
Grouped Key-Value Pairs
      │
      ▼
3. REDUCE
      │
      ▼
Final Output
```
### Stage 1: MAP

The Map stage is responsible for **splitting the dataset into smaller pieces** and converting each record into **key-value pairs**.

Each computer receives part of the dataset and processes it independently.

Think of the Map stage as assigning work to multiple employees.

Instead of one person reading an entire library, each person receives a few books.

#### What Happens During Mapping?
1. The dataset is split into smaller fragments.
2. Each fragment is sent to a different node.
3. Each node processes its fragment independently.
4. The Map function outputs key-value pairs.

#### Example 1: Word Count
Suppose we have the sentence:

```text
Lion Tiger Lion Elephant Lion Tiger
```
The Map function processes one word at a time.

#### Map Output

```text
(Lion,1)

(Tiger,1)

(Lion,1)

(Elephant,1)

(Lion,1)

(Tiger,1)
```

Notice that every word becomes a **key**, and every occurrence receives the value **1**.

Input:

```text
Laptop
Phone
Laptop
Tablet
Phone
Laptop
```

Map Output

```text
(Laptop,1)

(Phone,1)

(Laptop,1)

(Tablet,1)

(Phone,1)

(Laptop,1)
```

#### Why Key-Value Pairs?
Key-value pairs make data easy to:

- Sort
- Group
- Aggregate
- Analyze

Almost every Big Data framework processes information using key-value structures.

### Stage 2: SHUFFLE
The Shuffle stage automatically sorts and groups all identical keys together.

Developers **do not usually write Shuffle code**.

The MapReduce framework performs this step automatically.

#### Why is Shuffling Necessary?

Imagine three different computers processed three different files.

Node 1

```text
(Lion,1)

(Tiger,1)
```

Node 2

```text
(Lion,1)

(Giraffe,1)
```

Node 3

```text
(Giraffe,1)

(Lion,1)
```
Without shuffling:

The Reduce stage would never know that multiple "Lion" records belong together.

#### Shuffle Output

```text
Lion → [1,1,1]

Tiger → [1]

Giraffe → [1,1]
```

Notice:

All Lion values are now together.

All Tiger values are together.

All Giraffe values are together.

This grouping prepares the data for aggregation.

### Stage 3: REDUCE
The Reduce stage summarizes all values associated with the same key.

The Reduce function decides **how the values should be combined**.

Common operations include:

- Sum
- Average
- Maximum
- Minimum
- Count

#### Word Count Example

Shuffle Output

```text
Lion → [1,1,1]

Tiger → [1,1]

Elephant → [1]
```

Reduce Output

```text
Lion → 3

Tiger → 2

Elephant → 1
```

The Reduce function simply adds the numbers.

#### Online Shopping Example
Input after Shuffle

```text
Laptop → [1,1,1]

Phone → [1,1]

Tablet → [1]
```

Reduce Output

```text
Laptop → 3 Sales

Phone → 2 Sales

Tablet → 1 Sale
```

Now Amazon knows exactly how many units of each product were sold.

#### Complete MapReduce Workflow

```text
Raw Dataset
      │
      ▼
Split Into Chunks
      │
      ▼
Map Function
      │
      ▼
Key-Value Pairs
      │
      ▼
Shuffle
(Group Similar Keys)
      │
      ▼
Reduce Function
      │
      ▼
Final Results
```

# MapReduce Pseudocode

```python
# Map Function
def map(document):
    for word in document.split():
        emit(word, 1)


# Reduce Function
def reduce(key, values):
    emit(key, sum(values))
```

### Job Tracker and Task Tracker

MapReduce requires coordination to ensure all tasks are completed correctly.

#### Job Tracker (Master Node)

The Job Tracker is responsible for:

- Receiving the job.
- Splitting it into map and reduce tasks.
- Assigning tasks to worker nodes.
- Monitoring task progress.
- Reassigning failed tasks.

#### Task Tracker (Worker Nodes)
Task Trackers are responsible for:

- Executing Map tasks.
- Executing Reduce tasks.
- Reporting progress to the Job Tracker.
