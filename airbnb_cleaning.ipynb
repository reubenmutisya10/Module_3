```python
import pandas as pd

df = pd.read_csv("listings.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Check the columns we need for the Airbnb investment analysis

important_columns = [
    "id",
    "host_id",
    "host_is_superhost",
    "host_listings_count",
    "neighbourhood_cleansed",
    "latitude",
    "longitude",
    "property_type",
    "room_type",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "amenities",
    "price",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "review_scores_rating",
    "availability_365",
    "instant_bookable",
    "estimated_occupancy_l365d",
    "estimated_revenue_l365d"
]

df[important_columns].head()
```

```python
# Check duplicate listing IDs
print("Duplicate listing IDs:", df["id"].duplicated().sum())

# Check missing values in our important columns
missing = df[important_columns].isnull().sum()
missing_percent = (missing / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    "Missing Values": missing,
    "Missing %": missing_percent
})

missing_summary.sort_values("Missing %", ascending=False)
```

```python
# Create a clean dataframe containing the fields needed for Tableau
airbnb = df[important_columns].copy()

# Remove instant_bookable because it is 100% missing
airbnb = airbnb.drop(columns=["instant_bookable"])

# Replace missing reviews per month with 0
airbnb["reviews_per_month"] = airbnb["reviews_per_month"].fillna(0)

print("Rows:", airbnb.shape[0])
print("Columns:", airbnb.shape[1])

airbnb.head()
```

```python
# Convert price from text to numeric
airbnb["price_clean"] = (
    airbnb["price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

airbnb["price_clean"] = pd.to_numeric(
    airbnb["price_clean"],
    errors="coerce"
)

airbnb[["price", "price_clean"]].head(10)
```

```python
# Examine the main investment metrics for unusual values

metrics = [
    "price_clean",
    "estimated_revenue_l365d",
    "estimated_occupancy_l365d"
]

airbnb[metrics].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
).round(2)
```

```python
# View the 15 most expensive listings

airbnb[
    [
        "id",
        "neighbourhood_cleansed",
        "property_type",
        "room_type",
        "accommodates",
        "bedrooms",
        "price_clean",
        "estimated_occupancy_l365d",
        "estimated_revenue_l365d"
    ]
].sort_values(
    "price_clean",
    ascending=False
).head(15)
```

```python
# Calculate the 99th percentile price threshold
price_p99 = airbnb["price_clean"].quantile(0.99)

print("99th percentile price:", round(price_p99, 2))

# Flag extreme prices
airbnb["price_outlier"] = airbnb["price_clean"] > price_p99

print("\nPrice outliers:", airbnb["price_outlier"].sum())
print("Regular listings:", (~airbnb["price_outlier"]).sum())
```

```python
# Convert estimated occupied nights into an occupancy percentage
airbnb["estimated_occupancy_rate"] = (
    airbnb["estimated_occupancy_l365d"] / 365
) * 100

airbnb[
    [
        "estimated_occupancy_l365d",
        "estimated_occupancy_rate"
    ]
].describe().round(2)
```

```python
print("Rows:", airbnb.shape[0])
print("Columns:", airbnb.shape[1])

print("\nDuplicate IDs:", airbnb["id"].duplicated().sum())

print("\nMissing values:")
print(airbnb.isnull().sum().sort_values(ascending=False))

print("\nPrice statistics:")
print(airbnb["price_clean"].describe())

print("\nOccupancy rate statistics:")
print(airbnb["estimated_occupancy_rate"].describe())

print("\nNeighbourhoods:")
print("Number of neighbourhoods:", airbnb["neighbourhood_cleansed"].nunique())
print(airbnb["neighbourhood_cleansed"].value_counts())
```

```python
#Export cleaned Nairobi Airbnb dataset for Tableau

airbnb.to_csv("nairobi_airbnb_clean.csv", index=False)

print("Export complete!")
print("File: nairobi_airbnb_clean.csv")
print("Rows:", airbnb.shape[0])
print("Columns:", airbnb.shape[1])
```
