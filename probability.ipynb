# Probability and Distributions

# Part 1: Introduction to Probability 

## What is Probability?

Probability is the mathematics of uncertainty.

It measures **how likely an event is to happen**, with values ranging from:

- **0** → Impossible
- **1** → Certain

or

- **0%** → Impossible
- **100%** → Guaranteed

### Probability Formula

The probability of an event \(A\) occurring is:

$$
P(A) = \frac{\text{Number of favorable outcomes}}{\text{Total number of possible outcomes}}
$$

Where:

- \(P(A)\) = Probability of event \(A\)
- \(0 \le P(A) \le 1\)
---

## Simple Example

You roll a fair six-sided die.

What is the probability of rolling a 3?

- Favorable outcomes = 1
- Total outcomes = 6

Therefore,

P(3)=1/6 ≈ 0.167 (16.7%)

---

## Real-World Example

Imagine it rains on average **20 days every 100 days**.

Probability of rain tomorrow:

P(Rain)=20/100 = 0.20 = 20%

Meteorologists use historical weather data to estimate this probability.

## Probability vs Statistical Inference

### Probability

Starts with a known process and predicts possible outcomes.

Example:

"If a coin is fair, what is the probability of getting Heads?"

---

### Inference

Starts with observed data and tries to understand the underlying process.

Example:

"We observed many coin flips.

Is this coin actually fair?"

---

Think of them as:

Probability → Forward reasoning

Inference → Backward reasoning

# Part 3: Probability Distributions

## What is a Probability Distribution?

A probability distribution describes **how probabilities are assigned across all possible values of a variable.**

It tells us:

- Which values occur often
- Which values occur rarely

---

## Example

Imagine recording students' heights.

Most students have average heights.

Very few are extremely short.

Very few are extremely tall.

The data naturally forms a pattern called a **distribution**.

---

## Types of Probability Distributions

### Discrete Distribution

Deals with countable values.

Examples:

- Number of customers
- Number of goals
- Number of emails

---

### Continuous Distribution

Deals with measurable values.

Examples:

- Height
- Weight
- Temperature
- Income
- Exam scores

The Normal Distribution belongs here.

---

# Part 4: Normal Distribution 

## What is the Normal Distribution?

The Normal Distribution is a continuous probability distribution that forms a **bell-shaped curve**.

Characteristics:

- Symmetrical
- Bell-shaped
- Mean at the center
- Most values cluster near the average
- Fewer values occur at the extremes

---

## Examples

- Student exam scores
- Human heights
- IQ scores
- Blood pressure
- Manufacturing measurements

---

## Important Terms

### Mean (μ)

The average.

Example:

Average exam score = 70

---

### Standard Deviation (σ)

Measures how spread out the data is.

Small standard deviation:

Students scored similarly.

Large standard deviation:

Scores vary widely.

---

## Visual Interpretation

Imagine exam scores:

```
40 50 60 70 80 90 100

             ^
           Mean
```

Most students score close to 70.

Few score 40 or 100.

---

# Standard Normal Distribution

The Standard Normal Distribution is simply a Normal Distribution where:

Mean = 0

Standard deviation = 1



Instead of using original values, we convert everything into **Z-scores**.

---

## Z-score Formula

To standardize any value:

$$
z=\frac{x-\mu}{\sigma}
$$

where:

- \(x\) = observed value
- \(\mu\) = mean
- \(\sigma\) = standard deviation

---

## Example

Exam scores

Mean = 75

Standard deviation = 5

Student score = 85

$$
z=\frac{85-75}{5}=2
$$

Interpretation:

The student scored **2 standard deviations above the average.**

---

## Why Convert to Z-Scores?

Z-scores allow us to compare different datasets.

Example:

Student A:

90 in Mathematics

Student B:

170 cm height

Different units.

After converting to Z-scores, comparisons become meaningful.

# Part 5: Empirical Rule

The Empirical Rule explains how data is distributed in a Normal Distribution.

---

## 68–95–99.7 Rule

Approximately:

- **68%** of data lies within **±1σ**
- **95%** lies within **±2σ**
- **99.7%** lies within **±3σ**



---

## Example

Mean salary = KES 80,000

Standard deviation = KES 10,000

Then:

68%

KES 70,000 – 90,000

95%

KES 60,000 – 100,000

99.7%

KES 50,000 – 110,000

# Part 6: t-Distribution 
The t-distribution looks like the Normal Distribution but has **heavier tails**.

Use it when:

- Sample size is small
- Population standard deviation is unknown

Example:

Surveying only **15 customers**.

Since the sample is small, use the t-distribution instead of the Normal Distribution.
